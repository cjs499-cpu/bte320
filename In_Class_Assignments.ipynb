```python
number1 = int(input("Give Input 1: "))
number2 = int(input("Give Input 2: "))
operator = str(input("Give funtion"))

if operator == "+":
    number3 = number1 + number2
elif operator == "-":
    number3 = number1 - number2
elif operator == "*":
    number3 = number1 * number2
elif operator == "/":
    number3 = number1 / number2

print(number3)
```
