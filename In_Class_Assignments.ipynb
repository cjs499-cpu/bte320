```python
def rev(n):
    if len(n) == 1:
        return n
    else:
        return n[-1] + rev(n[:-1])

# Test the function
rev('12345')
```

```python
def roi(cost, revenue):
    assert isinstance(cost, (int, float)), "cost must be int or float"
    assert isinstance(revenue, (int, float)), "revenue must be int or float"
    assert cost != 0, "cost cannot be zero"
    return (revenue - cost) / cost

def loan_payment(principal, annual_rate, months):
    assert isinstance(principal, (int, float)), "principal must be a number"
    assert isinstance(annual_rate, (int, float)), "annual rate must be a number"
    assert isinstance(months, (int, float)), "months must be a number"

    if principal <= 0:
        raise ValueError("principal must be more than 0")
    if months <= 0:
        raise ValueError("months must be more than 0")
    if annual_rate < 0:
        raise ValueError("rate cannot be negative")

    r = annual_rate / 12

    if r != 0:
        pmt = r * principal / (1 - (1 + r) ** (-months))
    else:
        pmt = principal / months

    return pmt

try:
    principal = float(input("Enter principal: "))
    annual_rate = float(input("Enter annual rate: "))
    months = int(input("Enter months: "))
    print(loan_payment(principal, annual_rate, months))
except ValueError as e:
    print(e)
```

```python
class Student:
    def __init__(self, name, number):
        self.__name = name
        self.number = number
        self.__courses = []

    def get_name(self):
        return self.__name

    def get_courses(self):
        return self.__courses

    def set_name(self, newName):
        assert isinstance(newName, str)
        self.__name = newName

    def __len__(self):
        return len(self.__courses)

    def enroll(self, newCourse):
        if newCourse not in self.__courses:
            self.__courses.append(newCourse)
        else:
            print(f"Student {self.__name} is already enrolled in {newCourse}.")


# Usage
```

```python
def getValues(n):
    values = []
    for i in range(n):
        while True:
            try:
                val = float(input(f"Enter value {i+1}: "))
                values.append(val)
                break
            except ValueError:
                print("Invalid input, please enter a number")
    return values

def calcAverage(alist):
    return sum(alist) / len(alist)

n = int(input("How many values to enter? "))
values = getValues(n)
trimmed = values[3:]

if len(trimmed) > 0:
    print(f"Average (excluding first 3): {calcAverage(trimmed):.2f}")
else:
    print("Not enough values to calculate average")
```

```python
class Greeting:
    def __init__(self, first, last):
        self.first = first
        self.last = last

    def display(self, formal):
        if formal:
            print(f"{self.last}, {self.first}")
        else:
            print(self.first)

    def __str__(self):
        return f"{self.first} {self.last}"

g = Greeting("John", "Doe")
g.display(True)
g.display(False)
print(g)
```

```python
import math

class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def distance(self, other):
        return math.sqrt((self.x - other.x)**2 + (self.y - other.y)**2)

class Rectangle:
    def __init__(self, p1, p2, p3, p4):
        self.p1 = p1
        self.p2 = p2
        self.p3 = p3
        self.p4 = p4
        self.width = self.calc_width()
        self.height = self.calc_height()

    def calc_width(self):
        return self.p1.distance(self.p2)

    def calc_height(self):
        return self.p1.distance(self.p4)

    def calc_area(self):
        return self.width * self.height

p1 = Point(0, 0)
p2 = Point(4, 0)
p3 = Point(4, 3)
p4 = Point(0, 3)

rect = Rectangle(p1, p2, p3, p4)
print(f"Width: {rect.calc_width()}")
print(f"Height: {rect.calc_height()}")
print(f"Area: {rect.calc_area()}")
```

```python
class Employee:
    def __init__(self, name, wage, hours):
        self.name = name
        self.__wage = wage
        self.hours = hours

    def get_wage(self):
        return self.__wage

    def set_wage(self, new_wage):
        if new_wage < 0:
            print("Error: wage cannot be negative!")
        else:
            self.__wage = new_wage

    def calculate_pay(self):
        return self.__wage * self.hours

    def display_info(self):
        print(f"Name: {self.name}")
        print(f"Hourly Wage: ${self.get_wage():.2f}")
        print(f"Hours Worked: {self.hours}")
        print(f"Weekly Pay: ${self.calculate_pay():.2f}")


class Manager(Employee):
    def __init__(self, name, wage, hours, bonus):
        super().__init__(name, wage, hours)
        self.bonus = bonus

    def calculate_pay(self):
        return super().calculate_pay() + self.bonus

    def display_info(self):
        super().display_info()
        print(f"Bonus: ${self.bonus:.2f}")
        print(f"Total Weekly Pay: ${self.calculate_pay():.2f}")


emp = Employee("John Smith", 15.00, 40)
mgr = Manager("Jane Doe", 25.00, 40, 500.00)

print("--- Regular Employee ---")
emp.display_info()
print("\n--- Manager ---")
mgr.display_info()
```
