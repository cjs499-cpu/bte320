```python
sales = float(input("Enter sales value ($): "))
tax_rate = float(input("Enter tax rate (%): "))

cogs = 0.4 * sales
gross_profit = sales - cogs
taxes = gross_profit * tax_rate / 100
net = gross_profit - taxes

print(f"Sales: ${sales}")
print(f"Cost of Goods Sold: ${cogs}")
print(f"Gross Profit: ${gross_profit}")
print(f"Taxes: ${taxes}")
print(f"Net Income: ${net}")
```

```python
number1 = int(input("Give Input 1: "))
number2 = int(input("Give Input 2: "))
operator = input("Give funtion")

if operator == '+':
    number3 = number1 + number2
elif operator == '-':
    number3 = number1 - number2
elif operator == '*':
    number3 = number1 * number2
elif operator == '/':
    number3 = number1 / number2
elif operator == '%':
    number3 = number1 % number2
elif operator == '**':
    number3 = number1 ** number2
else:
    print("Invalid operator")
    number3 = None # Or handle error differently

if number3 is not None:
    print(number3)
```
